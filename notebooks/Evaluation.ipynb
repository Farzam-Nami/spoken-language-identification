# Integrated Evaluation of Spoken-Language Identification

Load the classification and clustering summary CSVs produced by [Classification](Classification.ipynb) and [Clustering](Clustering.ipynb). The included CSVs also contain metrics extracted from the saved notebook outputs, so this evaluation can be viewed without the raw audio. See [experimental analysis](../docs/experimental_analysis.md).

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

cls_csv = Path("../results/classification_results_summary.csv")
clu_csv = Path("../results/clustering_results_summary.csv")

if not cls_csv.exists():
    raise FileNotFoundError("classification_results_summary.csv not found. Run Classification.ipynb first.")
if not clu_csv.exists():
    raise FileNotFoundError("clustering_results_summary.csv not found. Run Clustering.ipynb first.")

df_cls = pd.read_csv(cls_csv)
df_clu = pd.read_csv(clu_csv)

print("Loaded classification results:", cls_csv.resolve())
display(df_cls)

print("Loaded clustering results:", clu_csv.resolve())
display(df_clu)
```

## Overall findings

The supervised results are strong on the fixed clip-level holdout (KNN: 98.48% accuracy, 130/132 correct). Unsupervised groups are less aligned with the four language labels (K-Means purity 72.06%; hierarchical purity 28.40%). These scores answer different questions: classification uses training labels and a six-component PCA representation, whereas clustering explores the full standardized feature space without labels. The custom silhouette calculation differs from the conventional definition, and the two clustering models use different k values. The current evidence does not establish generalization to unseen narrators, books or recording sources. A grouped split and repeated evaluation would be valuable future checks.
